```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

```python
model = keras.Sequential([
    layers.Dense(100, activation='elu', kernel_initializer='he_normal', input_shape=(784, )),
    layers.Dense(100, activation="elu", kernel_initializer="he_normal"),
    layers.Dense(100, activation="elu", kernel_initializer="he_normal"),
    layers.Dense(100, activation="elu", kernel_initializer="he_normal"),
    layers.Dense(100, activation="elu", kernel_initializer="he_normal"),

    layers.Dense(10, activation="softmax")
])
```

```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()
```

```python
x_train = x_train.reshape(-1, 784) / 255.0
x_test = x_test.reshape(-1, 784) / 255.0
```

```python
mask_train = y_train < 5
mask_test = y_test < 5
```

```python
x_train = x_train[mask_train]
y_train = y_train[mask_train]
x_test = x_test[mask_test]
y_test = y_test[mask_test]
```

```python
model.compile(loss="sparse_categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
```

```python
early_stopping = keras.callbacks.EarlyStopping(monitor="val_loss", patience=5, restore_best_weights=True)
```

```python
checkpoint_cb = keras.callbacks.ModelCheckpoint("mnist_model.keras", save_best_only=True)
```

```python
history = model.fit(x_train, y_train, epochs = 30, validation_split=0.1, callbacks=[early_stopping, checkpoint_cb], batch_size=32)
```

```python
test_loss, test_acc = model.evaluate(x_test, y_test)
print("Test accuracy (digits 0–4):", test_acc)
```

```python
def build_model(n_hidden=5, n_neurons=100, learning_rate=1e-3):
  model = keras.Sequential()
  model.add(keras.layers.Input(shape=(784, )))

  for _ in range(n_hidden):
    model.add(keras.layers.Dense(n_neurons, activation="elu", kernel_initializer="he_normal"))

  model.add(keras.layers.Dense(5, activation="softmax"))

  optimizer = keras.optimizers.Adam(learning_rate=learning_rate)
  model.compile(loss="sparse_categorical_crossentropy", optimizer=optimizer, metrics=["accuracy"])

  return model
```

```python
from tensorflow.keras.layers import Input, Dense, Concatenate
from tensorflow.keras.models import Model

input1 = Input(shape=(784,))
input2 = Input(shape=(784,))

def build_branch():
    x = Dense(100, activation='elu', kernel_initializer='he_normal')(input1)
    for _ in range(4):
        x = Dense(100, activation='elu', kernel_initializer='he_normal')(x)
    return x

from tensorflow.keras import Sequential
branch = Sequential([
    Dense(100, activation='elu', kernel_initializer='he_normal'),
    Dense(100, activation='elu', kernel_initializer='he_normal'),
    Dense(100, activation='elu', kernel_initializer='he_normal'),
    Dense(100, activation='elu', kernel_initializer='he_normal'),
    Dense(100, activation='elu', kernel_initializer='he_normal')
])

out1 = branch(input1)
out2 = branch(input2)

merged = Concatenate()([out1, out2])
hidden = Dense(10, activation='elu')(merged)
output = Dense(1, activation='sigmoid')(hidden)

model = Model(inputs=[input1, input2], outputs=output)
```
